```python
import yfinance as yf

# Tesla
tesla = yf.Ticker("TSLA")
tesla_stock_data = tesla.history(period="5y")

# first 5 rows
tesla_stock_data.head()
```

```python
# Extract Tesla revenue data using yfinance
tesla_financials = tesla.financials
tesla_revenue = tesla_financials.loc["Total Revenue"]

# Display revenue data
print("Title: Tesla Revenue Data")
print("Caption: Tesla revenue data extracted using yfinance, as web scraping was unsuccessful.")
tesla_revenue
```

```python
# Extract GameStop stock data
gamestop = yf.Ticker("GME")
gamestop_stock_data = gamestop.history(period="5y")

# Display first 5 rows
print("Title: GameStop Stock Data")
print("Caption: The dataset below shows GameStop's stock price history over the last 5 years.")
gamestop_stock_data.head()
```

```python
# Extract GameStop revenue data using yfinance
gamestop_financials = gamestop.financials
gamestop_revenue = gamestop_financials.loc["Total Revenue"]

# Display revenue data
print("Title: GameStop Revenue Data")
print("Caption: GameStop revenue data extracted using yfinance, as web scraping was unsuccessful.")
gamestop_revenue
```

```python
import matplotlib.pyplot as plt

# Function to plot stock price vs revenue
def plot_stock_vs_revenue(stock_data, revenue_data, company_name):
    fig, ax1 = plt.subplots(figsize=(12, 6))
    ax1.set_title(f"{company_name} Stock Price vs Revenue Over Time")

    ax1.set_xlabel("Date")
    ax1.set_ylabel("Stock Price (USD)", color='tab:blue')
    ax1.plot(stock_data.index, stock_data["Close"], label="Stock Price", color='tab:blue')
    ax1.tick_params(axis='y', labelcolor='tab:blue')

    ax2 = ax1.twinx()
    ax2.set_ylabel("Revenue (in billions)", color='tab:red')
    ax2.plot(revenue_data.index, revenue_data.values / 1e9, label="Revenue", color='tab:red')
    ax2.tick_params(axis='y', labelcolor='tab:red')

    fig.tight_layout()
    plt.figtext(0.5, -0.02, f"Figure: {company_name} stock price and revenue trend comparison over the last 5 years.", wrap=True, horizontalalignment='center', fontsize=10)
    plt.show()

# Plot Tesla stock vs revenue
print("Title: Tesla Stock vs Revenue Chart")
print("Caption: The graph below visualizes Tesla's stock price and revenue over the last 5 years.")
plot_stock_vs_revenue(tesla_stock_data, tesla_revenue, "Tesla")
```

```python
# Plot GameStop stock vs revenue
print("Title: GameStop Stock vs Revenue Chart")
print("Caption: The graph below visualizes GameStop's stock price and revenue over the last 5 years.")
plot_stock_vs_revenue(gamestop_stock_data, gamestop_revenue, "GameStop")
```
